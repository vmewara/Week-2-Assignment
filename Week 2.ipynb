```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_excel('employee_data_messy.xlsx')
df=pd.DataFrame(data)
print(df)
```

```python
df['Gender'] = df['Gender'].replace({
    'Male': 'M',
    'Female': 'F'
})
```

```python
df
```

```python
df['Gender'].isnull().sum()
```

```python
df['Gender']=df['Gender'].replace('Other','O')
df
```

```python
df['Gender']=df['Gender'].fillna(df['Gender'].mode()[0])
df['Gender']
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['Employee_ID']).copy()
```

```python
df.isnull().sum()
```

```python
df
```

```python
df[['Age', 'Salary', 'Performance_Rating']].describe()
```

```python
df['Age']=pd.to_numeric(df['Age'],errors='coerce')
df['Age']=df['Age'].fillna(df['Age'].median())
df['Age']
```

```python
df.isnull().sum()
```

```python
df['Salary']
```

```python
df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')
min_salary = df['Salary'].min()
min_salary
```

```python
max_salary=df['Salary'].max()
max_salary
```

```python
df['Salary'] = df['Salary'].replace(-1, np.nan)
```

```python
min_salary = df['Salary'].min()
min_salary
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
df['Salary']
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['Full_Name']).copy()
df
```

```python
df['Performance_Rating'] = pd.to_numeric(df['Performance_Rating'], errors='coerce')
df['Performance_Rating'] = df['Performance_Rating'].fillna(df['Performance_Rating'].median())
df['Performance_Rating']
```

```python
df.isnull().sum()
```

```python
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
df['Department']
```

```python
df = df.dropna(subset=['Hire_Date'])
```
